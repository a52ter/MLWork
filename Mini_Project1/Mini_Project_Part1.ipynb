**4. Perform EDA and Visualization**

**1.** Use **histograms** to show the frequency of flipper length and that of body mass.

**Rational:** Both columns contain continuous numerical data, making them suitable for the x-axis of a histogram.  We also want to compare the distribution patterns of the two columns. We use separate histograms for the three species as the distributions depend on the species.

**Insights:**  The histograms show approximately symmetric distributions for both flipper length and body mass within each species. There are also a few potential outliers.

**2.** Use a **scatter plot** to show the relationship between flipper length and body mass.

**Rational:** The corr() function shows a strong correlation between the two columns, making a scatter plot appropriate for visualizing their relationship.

**Insights:** The scatter plot shows a strong positive correlation between flipper length and body mass.

**3.** Use **bar plots** to show the average flipper length by species and the average body mass by species.

**Rational:** The average is calculated for each species category. In the bar plots, the x-axis represents species, while the y-axis represents the corresponding average values.

**Insights:** The bar plots compare the average flipper length and average body mass among the three species: Gentoo, Adelie, and Chinstrap. The plots show that Gentoo has greater average flipper length and body mass than the other two species. Adelie and Chinstrap have relatively similar average body masses. Adelie has a slightly smaller average flipper length than Chinstrap.





```python
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(12,8))

def plot_subplot(axes, species):
  #flipper_len = df['flipper_length_mm']
  flipper_len = df[df['species']==species]['flipper_length_mm']

  axes[0].hist(flipper_len, bins=50, edgecolor='black')
  #axes[0].set_title('Distribution of Flipper Length')
  axes[0].set_xlabel(species + ':  Flipper Length (mm)')
  axes[0].set_ylabel('Frequency')
  axes[0].grid(True, linestyle="--", alpha=0.3)
  axes[0].set_xlim(170,240)

  #body_mass = df['body_mass_g']
  body_mass = df[df['species']==species]['body_mass_g']

  axes[1].hist(body_mass, bins=50, edgecolor='black')
  #axes[1].set_title('Distribution of Body Mass')
  axes[1].set_xlabel(species + ':  Body Mass (g)')
  axes[1].set_ylabel('Frequency')
  axes[1].grid(True, linestyle="--", alpha=0.3)
  axes[1].set_xlim(2500,6500)

plot_subplot(axes[0], 'Adelie')
plot_subplot(axes[1], 'Chinstrap')
plot_subplot(axes[2], 'Gentoo')

fig.suptitle('Distribution of Flipper Length and Body Mass by Species',fontsize=16)
fig.supxlabel('Species')
fig.supylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
x = df['body_mass_g']
y = df['flipper_length_mm']

plt.scatter(x,y)
plt.title('Body Mass vs Flipper Length')
plt.xlabel('Body Mass (g)')
plt.ylabel('Flipper Length (mm)')
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

```python
ma = df.groupby('species')[['flipper_length_mm','body_mass_g']].mean()
ma
```

```python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12,5))

cat = ma.index

flipper_len = ma['flipper_length_mm'].values
axes[0].bar(cat, flipper_len)
axes[0].set_title('Average Flipper Length by Species')
axes[0].set_xlabel('Species')
axes[0].set_ylabel('Average Flipper Length (mm)')
axes[0].grid(True, linestyle="--", alpha=0.4)

body_mass = ma['body_mass_g'].values
axes[1].bar(cat, body_mass)
axes[1].set_title('Average Body Mass by Species')
axes[1].set_xlabel('Species')
axes[1].set_ylabel('Average Body Mass (g)')
axes[1].grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
